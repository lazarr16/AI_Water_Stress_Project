Exploratory Data Analysis (EDA)

Imports

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("../data/raw/nasa_climate_data.csv")

# Preview
df.head()
```

Dataset Overview

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

Summary Statistics

```python
df.describe()
```

Temperature comparison across cities

```python
avg_temp_city = (
    df.groupby("city")["avg_temperature"]
    .mean()
    .sort_values()
)

plt.figure(figsize=(10,5))
avg_temp_city.plot(kind="bar")

plt.title("Average Temperature by City")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)

plt.show()
```

Humidity comparison

```python
avg_humidity_city = (
    df.groupby("city")["avg_humidity"]
    .mean()
    .sort_values()
)

plt.figure(figsize=(10,5))
avg_humidity_city.plot(kind="bar")

plt.title("Average Humidity by City")
plt.ylabel("Humidity (%)")
plt.xticks(rotation=45)

plt.show()
```

Rainfall comparison

```python
rain_city = (
    df.groupby("city")["precipitation"]
    .mean()
    .sort_values()
)

plt.figure(figsize=(10,5))
rain_city.plot(kind="bar")

plt.title("Average Annual Rainfall by City")
plt.ylabel("Rainfall")
plt.xticks(rotation=45)

plt.show()
```

Correlation Matrix

```python
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

print(correlation)
```
